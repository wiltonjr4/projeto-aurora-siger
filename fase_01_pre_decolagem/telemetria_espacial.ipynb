# Sistema de Verificação Pré-Decolagem

Projeto acadêmico desenvolvido para simular a análise de telemetria de uma nave espacial.

O projeto contempla:

- organização e descrição da telemetria;
- algoritmo de verificação pré-decolagem;
- implementação em Python;
- análise energética;
- análise assistida por inteligência artificial;
- reflexão crítica sobre ética, impacto social e sustentabilidade.

> **Observação:** todos os valores, limites e condições utilizados neste projeto são fictícios e foram definidos exclusivamente para fins acadêmicos.


## 1. Organização e descrição da telemetria

Foram definidos os seguintes dados simulados:

| Parâmetro | Valor simulado | Faixa/condição segura |
|---|---:|---|
| Temperatura interna | 23 °C | 18 °C a 27 °C |
| Temperatura externa | 32 °C | -20 °C a 50 °C |
| Integridade estrutural | 1 | Deve ser 1 |
| Nível de energia | 87% | Mínimo de 70% |
| Pressão do tanque | 245 bar | 200 a 300 bar |
| Propulsão | 1 | Deve ser 1 |
| Navegação | 1 | Deve ser 1 |
| Comunicação | 1 | Deve ser 1 |
| Suporte à vida | 1 | Deve ser 1 |

Para os dados binários, `1` representa funcionamento adequado e `0` representa falha.


## 2. Algoritmo de verificação

A lógica utilizada é:

1. Ler os dados simulados da telemetria.
2. Comparar cada valor com sua faixa ou condição segura.
3. Registrar qualquer falha encontrada.
4. Se nenhuma falha for identificada, exibir **PRONTO PARA DECOLAR**.
5. Caso exista uma ou mais falhas, exibir **DECOLAGEM ABORTADA** e listar os problemas encontrados.


## 3. Script em Python — cenário seguro

```python
# ============================================
# SISTEMA DE VERIFICAÇÃO PRÉ-DECOLAGEM
# ============================================

# 1. Dados simulados da telemetria

temperatura_interna = 23
temperatura_externa = 32
integridade_estrutural = 1
nivel_energia = 87
pressao_tanque = 245

modulo_propulsao = 1
modulo_navegacao = 1
modulo_comunicacao = 1
modulo_suporte_vida = 1


# 2. Lista para armazenar possíveis problemas encontrados

falhas = []


# 3. Verificação dos parâmetros

if 18 <= temperatura_interna <= 27:
    print("Temperatura interna: OK")
else:
    print("Temperatura interna: FORA DA FAIXA")
    falhas.append("Temperatura interna fora da faixa segura.")


if -20 <= temperatura_externa <= 50:
    print("Temperatura externa: OK")
else:
    print("Temperatura externa: FORA DA FAIXA")
    falhas.append("Temperatura externa fora da faixa segura.")


if integridade_estrutural == 1:
    print("Integridade estrutural: OK")
else:
    print("Integridade estrutural: FALHA")
    falhas.append("Falha na integridade estrutural.")


if nivel_energia >= 70:
    print("Nível de energia: OK")
else:
    print("Nível de energia: INSUFICIENTE")
    falhas.append("Nível de energia abaixo do mínimo necessário.")


if 200 <= pressao_tanque <= 300:
    print("Pressão do tanque: OK")
else:
    print("Pressão do tanque: FORA DA FAIXA")
    falhas.append("Pressão do tanque fora da faixa segura.")


if modulo_propulsao == 1:
    print("Módulo de propulsão: OK")
else:
    print("Módulo de propulsão: FALHA")
    falhas.append("Falha no módulo de propulsão.")


if modulo_navegacao == 1:
    print("Módulo de navegação: OK")
else:
    print("Módulo de navegação: FALHA")
    falhas.append("Falha no módulo de navegação.")


if modulo_comunicacao == 1:
    print("Módulo de comunicação: OK")
else:
    print("Módulo de comunicação: FALHA")
    falhas.append("Falha no módulo de comunicação.")


if modulo_suporte_vida == 1:
    print("Módulo de suporte à vida: OK")
else:
    print("Módulo de suporte à vida: FALHA")
    falhas.append("Falha no módulo de suporte à vida.")


# 4. Resultado final

print("\n=== RESULTADO DA VERIFICAÇÃO ===")

if len(falhas) == 0:
    print("PRONTO PARA DECOLAR 🚀")
else:
    print("DECOLAGEM ABORTADA")

    print("\nProblemas encontrados:")

    for falha in falhas:
        print("-", falha)
```

## 4. Teste com cenário de falha

Para validar a lógica, também é possível simular valores fora das faixas seguras.  
No exemplo abaixo, foram inseridas quatro anomalias:

- temperatura interna de 31 °C;
- energia de 62%;
- pressão do tanque de 325 bar;
- falha no módulo de comunicação.


```python
# ============================================
# TESTE COM CENÁRIO DE FALHA
# ============================================

temperatura_interna_teste = 31
temperatura_externa_teste = 32
integridade_estrutural_teste = 1
nivel_energia_teste = 62
pressao_tanque_teste = 325

modulo_propulsao_teste = 1
modulo_navegacao_teste = 1
modulo_comunicacao_teste = 0
modulo_suporte_vida_teste = 1

falhas_teste = []

if not (18 <= temperatura_interna_teste <= 27):
    falhas_teste.append("Temperatura interna fora da faixa segura.")

if not (-20 <= temperatura_externa_teste <= 50):
    falhas_teste.append("Temperatura externa fora da faixa segura.")

if integridade_estrutural_teste != 1:
    falhas_teste.append("Falha na integridade estrutural.")

if nivel_energia_teste < 70:
    falhas_teste.append("Nível de energia abaixo do mínimo necessário.")

if not (200 <= pressao_tanque_teste <= 300):
    falhas_teste.append("Pressão do tanque fora da faixa segura.")

if modulo_propulsao_teste != 1:
    falhas_teste.append("Falha no módulo de propulsão.")

if modulo_navegacao_teste != 1:
    falhas_teste.append("Falha no módulo de navegação.")

if modulo_comunicacao_teste != 1:
    falhas_teste.append("Falha no módulo de comunicação.")

if modulo_suporte_vida_teste != 1:
    falhas_teste.append("Falha no módulo de suporte à vida.")

print("=== TESTE DE FALHA ===")

if len(falhas_teste) == 0:
    print("PRONTO PARA DECOLAR 🚀")
else:
    print("DECOLAGEM ABORTADA")
    print("\nProblemas encontrados:")
    for falha in falhas_teste:
        print("-", falha)
```

## 5. Análise energética

Para a simulação foram definidos:

- capacidade total: **500 kWh**;
- carga atual: **87%**;
- consumo estimado na decolagem: **120 kWh**;
- perdas energéticas: **5%**;
- consumo médio durante o voo: **50 kWh/h**.

O consumo médio durante o voo foi acrescentado à simulação para permitir a estimativa da autonomia em horas.


```python
# ============================================
# ANÁLISE ENERGÉTICA
# ============================================

capacidade_total = 500  # kWh
carga_atual = nivel_energia  # 87% da telemetria do cenário seguro
consumo_decolagem = 120  # kWh
taxa_perdas = 0.05  # 5%
consumo_voo_hora = 50  # kWh por hora

# Energia disponível de acordo com a carga atual
energia_armazenada = capacidade_total * (carga_atual / 100)

# Cálculo das perdas energéticas
perdas_energia = energia_armazenada * taxa_perdas

# Energia realmente disponível
energia_util = energia_armazenada - perdas_energia

# Energia restante depois da decolagem
energia_restante = energia_util - consumo_decolagem

print("\n=== ANÁLISE ENERGÉTICA ===")
print(f"Capacidade total: {capacidade_total} kWh")
print(f"Carga atual: {carga_atual}%")
print(f"Energia armazenada: {energia_armazenada:.2f} kWh")
print(f"Perdas energéticas: {perdas_energia:.2f} kWh")
print(f"Consumo na decolagem: {consumo_decolagem} kWh")

# Verifica se existe energia suficiente
if energia_restante > 0:

    autonomia_horas = energia_restante / consumo_voo_hora

    horas = int(autonomia_horas)
    minutos = int((autonomia_horas - horas) * 60)

    percentual_restante = (energia_restante / capacidade_total) * 100

    print(f"Energia restante (kWh): {energia_restante:.2f} kWh")
    print(f"Energia restante (%): {percentual_restante:.2f}%")
    print(f"Autonomia estimada: {autonomia_horas:.2f} horas")
    print(f"Autonomia aproximada: {horas}h {minutos}min")

else:
    print("Energia insuficiente para completar a decolagem.")
```
